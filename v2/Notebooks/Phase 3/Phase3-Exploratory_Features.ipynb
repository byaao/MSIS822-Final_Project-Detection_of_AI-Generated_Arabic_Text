### 2) f004 — Whitespace / Characters Ratio

**What:** How much of the text is made of spaces.

**Approach:**

```python
text = "الذكاء الاصطناعي رائع جدا"
whites = sum(ch.isspace() for ch in text)
ratio = whites / len(text)
ratio
```

**Result:** `0.12`

**Interpretation:** Higher ratio → shorter words or many small tokens; lower ratio → dense style.

### 3) f013 — Hapax Ratio

**What:** Fraction of words that appear exactly once.

**Approach:**


```python
from collections import Counter
tokens = "الذكاء الاصطناعي الذكاء هو المستقبل".split()
freq = Counter(tokens)
hapax_ratio = sum(1 for w,c in freq.items() if c==1) / len(tokens)
hapax_ratio
```

**Result:** `0.6`

**Interpretation:** Higher ratio → richer vocabulary; lower → repetitive language.

### 4) f022 — Entropy of Word Frequencies

**What:** Shannon entropy = how evenly word frequencies are distributed.

**Approach:**

```python
import math
p = [c/sum(freq.values()) for c in freq.values()]
H = -sum(pi*math.log2(pi) for pi in p)
H
```

**Result:** `1.92`

**Interpretation:** Higher H → diverse word usage; lower H → few words dominate.

### 12) f076 — Passive Sentences

**What:** Count of sentences containing any passive verb (`vox='p'`).

**Approach:**

```python
from camel_tools.disambig.mle import MLEDisambiguator
from camel_tools.tokenizers.word import simple_word_tokenize
import regex as re

mle = MLEDisambiguator.pretrained()

PASSIVE_HINT = re.compile(r'^[^\W\d_]\u064F[^\W\d_]\u0650[^\W\d_]\u064E$', re.UNICODE)

def has_passive_voice(tokens, disamb_words):
    # 1) morphology voice
    for w in disamb_words:
        for sc in getattr(w, 'analyses', []):
            if (sc.analysis.get('vox') or '').lower().startswith('p'):
                return True
    # 2) surface pattern fallback (works on fully diacritized Form I perfect)
    return any(PASSIVE_HINT.match(tok) for tok in tokens)

def count_passive_sentences(sentences):
    tokenized = [simple_word_tokenize(s) for s in sentences]
    disamb    = [mle.disambiguate(toks) for toks in tokenized]
    return sum(has_passive_voice(toks, dw) for toks, dw in zip(tokenized, disamb))

sentences = ["كُتِبَ الدرس.", "قرأ الطالب الدرس."]
count_passive_sentences(sentences)
```

**Result:** `1`

**Interpretation:** Higher → impersonal or formal writing style.

### 16) f088 — Semantic Similarity between Sentences

**What:** Average cosine similarity between consecutive sentence embeddings.

**Approach:**

```python
%pip install -U "transformers==4.36.2" "sentence-transformers==2.5.1" torch
```

```python
from sentence_transformers import SentenceTransformer
import torch, numpy as np
model = SentenceTransformer("Omartificial-Intelligence-Space/Arabic-arabert-all-nli-triplet")
sents = ["الذكاء الاصطناعي يتطور بسرعة.", "هذا المجال مهم جداً."]
emb = model.encode(sents, convert_to_tensor=True, normalize_embeddings=True)
np.mean([float(torch.nn.functional.cosine_similarity(emb[i], emb[i+1], dim=0)) for i in range(len(emb)-1)])
```

**Result:** `0.31`

**Interpretation:** Near 1 → sentences are semantically similar (cohesive); near 0 → topic shifts.

### 17) f097 — BERT Token Embedding Similarity

**What:** Mean cosine similarity of consecutive subword embeddings within a sentence (AraBERT).

**Approach:**

```python
from transformers import AutoTokenizer, AutoModel
import torch, numpy as np
tok = AutoTokenizer.from_pretrained("aubmindlab/bert-base-arabertv02")
mod = AutoModel.from_pretrained("aubmindlab/bert-base-arabertv02")
inputs = tok("كرة القدم كاس العالم", return_tensors="pt")
with torch.no_grad(): h = mod(**inputs).last_hidden_state.squeeze(0)
np.mean(torch.nn.functional.cosine_similarity(h[:-1], h[1:], dim=1).numpy())
```

**Result:** `0.55`

**Interpretation:** High → smooth token transitions; low → abrupt or code-switched text.
